```python
import pandas as pd
from sdv.metadata import SingleTableMetadata
from sdv.single_table import CTGANSynthesizer

# CSV dosyasını okur
csv_file = "data/lung_msk_mind_2020_clinical_data.csv"
data = pd.read_csv(csv_file)
data = data.fillna(0)  # Eksik verileri 0 ile doldurur

# Metadata oluşturuluyor
metadata = SingleTableMetadata()

# Veri setindeki her bir sütun için metadata'yı ekleyin
for column in data.columns:
    if data[column].dtype == 'object':  # Eğer sütun kategorikse
        metadata.add_column(column, sdtype='categorical')
    else:  # Eğer sütun sayısal bir veri içeriyorsa
        metadata.add_column(column, sdtype='numerical')

# CTGAN modelini oluştur
synthesizer = CTGANSynthesizer(metadata=metadata)

# Modeli eğit (Epoch sayısı değişebilir, örneğin 1000)
synthesizer.fit(data)

# 100 yeni sentetik veri üret
new_data = synthesizer.sample(100)

# Üretilen sentetik verileri görüntüle
print("Sentetik veri seti ilk 5 satırı:")
print(new_data.head())

# Veri setlerinin sütunlarının aynı olduğundan emin olalım
print("Orijinal veri sütunları:", data.columns)
print("Sentetik veri sütunları:", new_data.columns)

# Eğer sütunlar farklıysa, yeni veriyi orijinal veriyle uyumlu hale getirin
new_data = new_data[data.columns]

# Veriyi birleştirme
combined_data = pd.concat([data, new_data], ignore_index=True)

# Birleşik veriyi kaydetme
combined_data.to_csv('combined_data.csv', index=False)

# Yeni veriyi başarıyla kaydettik, ekrana bir mesaj yazdıralım
print("Veriler başarıyla birleştirildi ve 'combined_data.csv' olarak kaydedildi.")
```
